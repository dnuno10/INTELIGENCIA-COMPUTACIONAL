```python
import joblib
from skl2onnx import convert_sklearn
from skl2onnx.common.data_types import FloatTensorType, StringTensorType

model = joblib.load("/Users/dnuno/Documents/INTELIGENCIA-COMPUTACIONAL/EXPORT-MODEL/ENTRENAMIENTO/car_price_model.joblib")

initial_types = [
    ("Manufacture_Year", FloatTensorType([None, 1])),
    ("Engine_CC", FloatTensorType([None, 1])),
    ("Horsepower", FloatTensorType([None, 1])),
    ("Mileage_km_per_l", FloatTensorType([None, 1])),
    ("Brand", StringTensorType([None, 1])),
    ("Body_Type", StringTensorType([None, 1])),
    ("Fuel_Type", StringTensorType([None, 1])),
    ("Transmission", StringTensorType([None, 1]))
]

onnx_model = convert_sklearn(
    model,
    initial_types=initial_types,
    target_opset=12
)

with open("car_price_model.onnx", "wb") as f:
    f.write(onnx_model.SerializeToString())
```

```python
import onnxruntime as ort
import numpy as np

session = ort.InferenceSession("car_price_model.onnx")

input_data = {
    "Manufacture_Year": np.array([[2020]], dtype=np.float32),
    "Engine_CC": np.array([[2000]], dtype=np.float32),
    "Horsepower": np.array([[150]], dtype=np.float32),
    "Mileage_km_per_l": np.array([[14.5]], dtype=np.float32),
    "Brand": np.array([["Toyota"]]),
    "Body_Type": np.array([["SUV"]]),
    "Fuel_Type": np.array([["Petrol"]]),
    "Transmission": np.array([["Automatic"]])
}

pred = session.run(None, input_data)
print("Predicción ONNX:", pred)
```
